```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, np.pi/2, 200) # zenith angle, 0 to 90 deg
I_v = 70 # m^-2 s^-1 sr^-1
I = I_v * np.cos(theta)**2

plt.plot(np.degrees(theta), I)
plt.xlabel("Zenith angle (degrees)")
plt.ylabel(r"Intensity $I(\theta)$ (m$^{-2}$ s$^{-1}$ sr$^{-1}$)")
plt.title("Cosmic muon flux vs. zenith angle")
plt.tight_layout()
plt.savefig("../figures/01_flux_vs_angle.png", dpi=300, bbox_inches="tight")
plt.show()
```
