```python
import numpy as np
import matplotlib.pyplot as plt
a = np.linspace(0, np.pi/2, 200)
I_v = 70
I = I_v*np.cos(a)**2

plt.plot(np.degrees(a), I)

plt.xlabel("Zenith angle (degrees)")
plt.ylabel(r"Flux $I(\theta)$ (m$^{-2}$ s$^{-1}$ sr$^{-1}$)")
plt.title("Cosmic muon flux vs. zenith angle")

plt.tight_layout()
plt.savefig("junyi_flux_vs_angle.png", dpi=300, bbox_inches="tight")
plt.show()
```
